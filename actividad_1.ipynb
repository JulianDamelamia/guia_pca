>Leer la sección 14.5.1, “Principal Components”, del libro The Elements of Statistical Learning. Identificar la idea central del método (PCA) y cómo se relaciona la variabilidad de los datos con las componentes principales

En resumen, el método PCA consiste en aproximar una observación $x_i$ como una combinación lineal $promedio+detalle$. El "detalle" puede ser más o menos granular, dependiendo de cuántas componentes principales se tomen. Estas componentes se obtienen descomponiendo la matrix de observaciones por SVD y quedándose con las primeras $q$ columnas. Más explícitamente, si $X = UDV^T$, la aproximación de grado $q$ es $$x_i =\bar x +  \sum_{j=1}^{q}  \lambda_{ij} v_j$$ donde los $v_j$ son las direcciones de las componentes principales y $\lambda_{ij} = U_{ij}d_j$. Estas componentes tienen la propiedad de ser las $q$ componentes que más varianza de los datos capturan* de todas las combinaciones de $q$ componentes posibles. Por dar un ejemplo, si los datos fueran caras humanas, una cara $x_i$ se aproxima como $cara\_ promedio +\lambda_{i1} \times razgo_1 + \lambda_{i2} \times razgo_2 + ...$.

###### * La matriz de covarianza $X^TX$ al descomponer por SVD se escribe como $VD^2V^T$. Como $d_0\geq d_1 \geq ...\geq d_N$, la componente $V_1$ tiene asociada la varianza $d_1^2$ que es mayor a todas las demás, $V_2$ tiene asociada $d_2^2$ que es mayor a todas las demás direcciones ortogonales a $V_1$ y así.
